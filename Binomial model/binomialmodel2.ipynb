```python
import numpy as np
```

```python
#initialising variables

S0 = float(input("Please insert spot price: "))
K = float(input("Please insert strike price: "))
T = int(input("Please insert the time (in years) till option expiry: "))
u = float(input("Please insert an up factor by which the stock increases each period: "))
d = 1/u
r = float(input("Please specify a risk-free interest rate: "))
N = int(input("Please insert the number of steps in the binomial tree model: "))

def binomial_tree_model_two(S0,K,T,u,d,r,N):
    dt = T/N #this is the duration per step (needed for appropriate discounting periods)
    p = (np.exp(r*dt) - d) / (u - d) #risk-neutral probability
    discount = np.exp(-r*dt)

    #building the binomial tree

    #Stock values at final time-step
    S = np.zeros(N+1)
    S[0] = S0 * d**N
    print("Potential stock values at maturity: ")
    print(f"0. {S[0]:.2f}")
    for j in range(1,N+1):
        S[j] = S[j-1] * u/d
        print(f"{j}. {S[j]:.2f}")
    
    #initialise call option values at maturity
    print(f"\nPotential call option payoffs at maturity: (K={K}) ")
    C = np.zeros(N+1)
    for j in range(0,N+1):
        C[j] = max(0,S[j]-K)
        print(f"{j}. {C[j]:.2f}")
    
    #finding present value of call option using backward induction
    for i in range(N,0,-1):
        for j in range(i):
            C[j] = (p * C[j+1] + (1-p) * C[j]) * discount
    print(f"\n The value of the call option today is {C[0]:.2f}")
    

binomial_tree_model_two(S0,K,T,u,d,r,N)


    
```
