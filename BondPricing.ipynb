```python
import numpy as np
import pandas as pd

#Bond valuation 
#P = CF/(1+r)**t + F/(1+r)**t

Coupon = float(input("Coupon in 0.00: "))
YTM = float(input("Insert prevailing market rate (YTM% in 0.00): "))
Face = int(input("Insert face value in EUR: "))
T = int(input("Insert Time in years:" ))

#Needed lists to for summary table
Years = []
Payment = []
PVpayment = []
CFxT = []

# Calculating the sum of future cash flows
SumCF = 0

for i in range(1,T):
    CF = (Coupon * Face)/(1+YTM)**i
    SumCF = SumCF + CF
    cfxt = CF * i

    Years.append(i)
    Payment.append(Coupon*Face)
    PVpayment.append(CF)
    CFxT.append(cfxt)

#Final Cash Flow
FinalCF = Face + Coupon * Face
PVFinalCF = (FinalCF) / (1+YTM)**T
FinalCFxT = PVFinalCF * T

SumCF = SumCF + PVFinalCF

Years.append(T)
Payment.append(FinalCF)
PVpayment.append(PVFinalCF)
CFxT.append(FinalCFxT)

print(f"The value of the bond is EUR {SumCF:.2f}")

#If condition specifying bond's market value
if abs(SumCF - Face) < 0.01:
    print("The bond is trading at par.")
elif SumCF < Face:
    print("The bond is trading at a discount.")
elif SumCF > Face:
    print("The bond is trading at a premium.")

print("\nSummary Table: \n")
table = pd.DataFrame({
    "Year": Years,
    "Cash Flow": Payment,
    "Discounted CFs": PVpayment,
    "Cf * T": CFxT,
})

total_row = pd.DataFrame({
    "Year": ["Total"],
    "Cash Flow": [sum(Payment)],
    "Discounted CFs": [sum(PVpayment)],
    "Cf * T": [sum(CFxT)]
})
table = pd.concat([table, total_row],ignore_index= True)

print(table.round(2).to_string(index=False))
```

```python
# Computing Macaulay, Modified Duration and Convexity of the bond: 

macd = sum(CFxT) / SumCF 
print(f"The Macaulay duration is {macd:.2f} years.\nIt takes {macd:.2f} years to recover the bond's price through its cash flows. ")

modd = macd / (1+YTM)
print(f"\nThe Modified duration is equal to {modd:.2f}.")
print(f"This means that a 1% increase in market yields leads to a {modd:.2f}% decrease in the bond's price.")
```
