```python
import numpy as np
```

```python
#Implementing a binomial model that computes the price of a European Option, 
#based on a multi-step binomial tree.
```

```python
#Variable initialisation

S0 = 100
K = 100
u = 1.1
d = 1/u
r = 0.05
T = 0.5
N = 3 #Number of Time Steps
opttype = 'C'
```

```python
def binomial_tree_slow(S0,K,u,d,r,T,N,opttype='C'):
    dt = T/N
    p = (np.exp(r*dt)-d)/(u-d)
    disc = np.exp(-r*dt)

    #initialise asset prices at maturity
    S = np.zeros(N+1)
    S[0] = S0 * d**N
    for j in range(1,N+1):
        S[j] = S[j-1]*u/d

    #initialise option values at maturity
    C = np.zeros(N+1)
    for j in range(0,N+1):
        C[j] = max(0,S[j]-K)

    #backward induction
    for i in np.arange(N,0,-1):
        for j in range(0,i):
            C[j] = disc*(p*(C[j+1]) + (1-p)*C[j])

    return C[0]  
```
